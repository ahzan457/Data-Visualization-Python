```python
# ==========================================
# MODULE 03: DATA VISUALIZATION WITH PYTHON
# ==========================================

# 1. IMPORT LIBRARIES
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Enable inline plotting
%matplotlib inline
print("Libraries imported successfully!")

# 2. CREATE SAMPLE DATASET FOR VISUALIZATION
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [12000, 15000, 18000, 14000, 22000, 26000],
    'Expenses': [8000, 9000, 11000, 9500, 13000, 15000],
    'Category': [
        'Electronics',
        'Clothing',
        'Electronics',
        'Home',
        'Electronics',
        'Clothing',
    ],
    'Customer_Rating': [4.2, 4.5, 4.8, 3.9, 4.7, 4.9],
}

df = pd.DataFrame(data)
print('Dataset created successfully!')
print(df)

# 3. LINE CHART (Trends over time)
plt.figure(figsize=(8, 4))
plt.plot(
    df['Month'],
    df['Sales'],
    marker='o',
    color='b',
    linewidth=2,
    label='Sales',
)
plt.plot(
    df['Month'],
    df['Expenses'],
    marker='s',
    color='r',
    linewidth=2,
    label='Expenses',
)
plt.title('Monthly Sales vs Expenses Trend')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True)
plt.show()

# 4. BAR CHART (Comparing categories)
plt.figure(figsize=(8, 4))
sns.barplot(x='Month', y='Sales', data=df, palette='viridis', hue='Month', legend=False)
plt.title('Monthly Sales Bar Chart')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.show()

# 5. HISTOGRAM (Distribution of numerical data)
plt.figure(figsize=(8, 4))
sns.histplot(df['Sales'], bins=4, kde=True, color='purple')
plt.title('Sales Distribution Histogram')
plt.xlabel('Sales Range')
plt.ylabel('Frequency')
plt.show()

# 6. SCATTER PLOT (Relationship between two variables using Seaborn)
plt.figure(figsize=(8, 4))
sns.scatterplot(
    x='Sales',
    y='Customer_Rating',
    hue='Category',
    data=df,
    s=100,
    palette='deep',
)
plt.title('Sales vs Customer Rating Scatter Plot')
plt.xlabel('Sales ($)')
plt.ylabel('Customer Rating')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

# 7. PIE CHART (Proportion breakdown)
plt.figure(figsize=(6, 6))
category_sales = df.groupby('Category')['Sales'].sum()
plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    colors=['#ff9999', '#66b3ff', '#99ff99'],
    startangle=140,
)
plt.title('Sales Proportion by Category')
plt.show()
```
